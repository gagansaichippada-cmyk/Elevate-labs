```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python

df = pd.read_csv("C:\\Users\\UOS-A-04\\Documents\\elevate labs\\aiml\\Task_1_Data_Cleaning\\dataset\\archive (3)\\Titanic-Dataset.csv")

print("Dataset imported successfully!")
```

```python
# Complete dataset information

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))


print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nFirst 10 Records:")
display(df.head(10))
```

```python
# Fill missing Age values using Median

df["Age"] = df["Age"].fillna(df["Age"].median())

print("Missing Age values handled successfully!")

# Check missing values in Age
print("Missing Age:", df["Age"].isnull().sum())
```

```python
# Handle missing Cabin values

df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing Cabin values handled successfully!")

# Check missing values
print("Missing Cabin:", df["Cabin"].isnull().sum())

# Display Cabin values
print(df["Cabin"].head(10))

df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)

print("1. Embarked missing values handled!")
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Categorical Encoding

df = pd.get_dummies(
    df,
    columns=["Sex", "Embarked"],
    drop_first=True,
    dtype=int
)

print("Categorical Encoding Completed!")

print("\nUpdated Columns:")
print(df.columns.tolist())

print("\nFirst 5 Records:")
display(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

# Create scaler
scaler = MinMaxScaler()

# Apply normalization
df[["Age", "Fare"]] = scaler.fit_transform(df[["Age", "Fare"]])

print("Feature Scaling Completed!")

# Display scaled values
display(df[["Age", "Fare"]].head())

# Check minimum and maximum values
print("\nMinimum Values:")
print(df[["Age", "Fare"]].min())

print("\nMaximum Values:")
print(df[["Age", "Fare"]].max())
```

```python
# Outlier Detection Using Boxplot

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.boxplot(data=df[["Age", "Fare"]])

plt.title("Outlier Detection in Titanic Dataset")
plt.xlabel("Features")
plt.ylabel("Normalized Values")

plt.show()
```

```python
# Step 9: Detect Outliers Using IQR Method

# Select numerical columns
columns = ["Age", "Fare"]

for column in columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print("\nColumn:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
```

```python
# Step 10: Handle Outliers Using IQR Method

columns = ["Age", "Fare"]

for column in columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Cap outliers
    df[column] = df[column].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("Outliers handled successfully!")

# Verify outliers after treatment
for column in columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    count = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    print(f"{column} remaining outliers: {count}")
```
